Real-Time Video Frame Processing and Image Manipulation

The notebook demonstrates real-time video frame capture using OpenCV, applies multiple filtering techniques, and performs basic image manipulations.

Objectives:
Capture video frames from the default webcam for 5 seconds.
Apply image filters (Grayscale, Gaussian Blur, Edge Detection).
Save every 30th processed frame.
Perform rotation, cropping, and flipping on saved frames.
Maintain clear and well-documented code.

1. Import Required Libraries

```python
import cv2
import os
from datetime import datetime
import matplotlib.pyplot as plt
```

2. Real-Time Video Capture and Filtering

```python
import cv2
import os
import time

# Create output folder
os.makedirs("output_frames", exist_ok=True)

cap = cv2.VideoCapture(0)
frame_count = 0

# Record start time
start_time = time.time()

while True:
    x, frame = cap.read()
    if not x:
        break

    # Apply filters
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    blur = cv2.GaussianBlur(frame, (15, 15), 0)
    edges = cv2.Canny(frame, 100, 200)

    # Show frames
    cv2.imshow('Gray', gray)
    cv2.imshow('Blur', blur)
    cv2.imshow('Edges', edges)

    # Save every 30th frame
    if frame_count % 30 == 0:
        filename = f"output_frames/frame_{frame_count}.jpg"
        cv2.imwrite(filename, gray)  # Save grayscale version

    frame_count += 1

    # Stop after 5 seconds
    if time.time() - start_time > 5:
        break

cap.release()
cv2.destroyAllWindows()
```

3. Image Manipulation on Saved Frames

```python
img = cv2.imread("output_frames/frame_0.jpg")

# Rotate 90 degrees
rotated = cv2.rotate(img, cv2.ROTATE_90_CLOCKWISE)

# Crop (100:400, 100:400)
cropped = img[100:400, 100:400]

# Flip horizontally
flipped = cv2.flip(img, 1)

cv2.imwrite("output_frames/rotated.jpg", rotated)
cv2.imwrite("output_frames/cropped.jpg", cropped)
cv2.imwrite("output_frames/flipped.jpg", flipped)
```
